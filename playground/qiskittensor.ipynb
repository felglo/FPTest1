```python
def function(psi, operator, i, N):
    s = "bcdefghijklmnopqrstuvwxyz"
    s[:N]
    I = s[i]
    final = s[:i] + "a" + s[i + 1 : N]
    psi_tensor = np.reshape(psi, (2,) * N, order="F")
    tensor_result = np.einsum(f"a{I},{s[:N]}->{final}", operator, psi_tensor)
    psi_result = np.reshape(tensor_result, -1, order="F")
    return psi_result
```

```python
cnot_reshaped = np.reshape(cnot, (2,) * 4, order="F")

print(cnot_reshaped)
```

```python
def newfunction(psi, operator, i, j, N):
    s = "cdefghijklmnopqrstuvwxyz"
    s[:N]
    I = s[i]
    J = s[j]
    if i < j:
        final = s[:i] + "a" + s[i + 1 : j] + "b" + s[j + 1 : N]
    if i > j:
        final = s[:j] + "b" + s[j + 1 : i] + "a" + s[i + 1 : N]
    psi_tensor = np.reshape(psi, (2,) * N, order="F")
    tensor_result = np.einsum(f"ab{I}{J},{s[:N]}->{final}", operator, psi_tensor)
    psi_result = np.reshape(tensor_result, -1, order="F")
    return psi_result
```

```python
psi = np.array([0, 0, 0, 0, 1, 0, 0, 0])
```

```python
final = newfunction(psi=psi, operator=cnot_reshaped, i=0, j=2, N=3)
print(final)
```
